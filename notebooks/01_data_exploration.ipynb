# 01 - Data Exploration
Explore the CREMA-D dataset: class balance, sample counts, and a quick look at raw examples.

```python
import sys
sys.path.append('..')

from src.dataset import load_crema_d

dataset = load_crema_d()
print(dataset)
```

```python
print(dataset['train'][0])
```

```python
from collections import Counter

labels = dataset['train']['label']
counts = Counter(labels)
print(counts)
```

```python
import matplotlib.pyplot as plt

plt.bar(counts.keys(), counts.values())
plt.xlabel('Emotion')
plt.ylabel('Number of Samples')
plt.title('Emotion Distribution')
plt.show()
```
