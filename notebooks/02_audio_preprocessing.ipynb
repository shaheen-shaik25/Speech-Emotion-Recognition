# 02 - Audio Preprocessing
Demonstrate signal-processing steps: waveform inspection, resampling, Mel spectrograms, and MFCCs.

```python
import sys
sys.path.append('..')

from src.dataset import load_crema_d
from src.preprocessing import compute_mel_spectrogram, compute_mfcc
import librosa, librosa.display
import numpy as np
import matplotlib.pyplot as plt

dataset = load_crema_d()
example = dataset['train'][0]
audio = example['audio']['array']
sr = example['audio']['sampling_rate']
```

```python
plt.figure(figsize=(12, 4))
librosa.display.waveshow(audio, sr=sr)
plt.title('Speech Waveform')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.show()
```

```python
mel_db = compute_mel_spectrogram(audio, sr=sr)

plt.figure(figsize=(12, 5))
librosa.display.specshow(mel_db, sr=sr, x_axis='time', y_axis='mel')
plt.colorbar(format='%+2.0f dB')
plt.title('Mel Spectrogram')
plt.show()
```

```python
mfcc = compute_mfcc(audio, sr=sr)
print(mfcc.shape)

plt.figure(figsize=(12, 5))
librosa.display.specshow(mfcc, x_axis='time', sr=sr)
plt.colorbar()
plt.title('MFCC Features')
plt.show()
```

## Notes
The raw waveform is converted into a time-frequency representation using a Mel spectrogram, and into MFCCs, to analyze how frequency content varies over time. These features complement the learned Wav2Vec2 representations used later.
