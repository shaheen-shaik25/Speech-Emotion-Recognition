# 03 - Baseline Models
Establish baselines using traditional features (MFCC) with classical ML models before moving to Wav2Vec2 fine-tuning.

```python
import sys
sys.path.append('..')

import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, f1_score

from src.dataset import load_crema_d
from src.preprocessing import compute_mfcc
from src.config import LABELS
```

```python
dataset = load_crema_d()

def extract_features(split):
    X, y = [], []
    for example in split:
        audio = example['audio']['array']
        sr = example['audio']['sampling_rate']
        mfcc = compute_mfcc(audio, sr=sr)
        X.append(mfcc.mean(axis=1))  # simple mean-pooled MFCC feature vector
        y.append(example['label'])
    return np.array(X), np.array(y)

# NOTE: for a quick baseline, consider subsampling the dataset before running this on the full set.
X_train, y_train = extract_features(dataset['train'])
X_test, y_test = extract_features(dataset['test'])
```

```python
clf = RandomForestClassifier(n_estimators=200, random_state=42)
clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)

print('Accuracy:', accuracy_score(y_test, y_pred))
print('F1 (macro):', f1_score(y_test, y_pred, average='macro'))
print(classification_report(y_test, y_pred, target_names=LABELS))
```
